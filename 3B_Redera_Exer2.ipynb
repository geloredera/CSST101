```python
import numpy as np
import matplotlib.pyplot as plt

# Function to calculate joint probability
def joint_probability(p_A, p_B):
    return p_A * p_B  # Assuming independence

# Function to calculate marginal probability
def marginal_probability(p_A, p_B):
    return p_A + p_B - joint_probability(p_A, p_B)

# Function to calculate conditional probability
def conditional_probability(p_A, p_B):
    return joint_probability(p_A, p_B) / p_B if p_B > 0 else 0

# Simulate a decision-making process
def simulate_decision(success_prob, iterations=1000):
    successes = np.random.binomial(1, success_prob, iterations)
    return successes

# Calculate expected value of a decision
def expected_value(success_prob, reward_success=1000, penalty_failure=-500):
    return (success_prob * reward_success) + ((1 - success_prob) * penalty_failure)

# Bayesian inference function
def bayesian_inference(prior, likelihood, evidence):
    posterior = (likelihood * prior) / evidence
    return posterior

# Example usage
p_A = 0.3  # Probability of event A
p_B = 0.4  # Probability of event B

# Joint Probability Calculation
joint_prob = joint_probability(p_A, p_B)
print(f"Joint Probability of A and B: {joint_prob}")

# Marginal Probability Calculation
marginal_prob = marginal_probability(p_A, p_B)
print(f"Marginal Probability of A and B: {marginal_prob}")

# Conditional Probability Calculation
cond_prob = conditional_probability(p_A, p_B)
print(f"Conditional Probability of A given B: {cond_prob}")

# Simulate decision-making process
success_prob = 0.7
iterations = 1000
outcomes = simulate_decision(success_prob, iterations)
average_return = np.mean(outcomes * 1000 - (1 - outcomes) * 500)
print(f"Average return after {iterations} iterations: {average_return}")

# Bayesian Inference Example
prior_disease = 0.01  # Prior probability of having the disease
likelihood_positive_test = 0.9  # Likelihood of positive test given disease
evidence_positive_test = (likelihood_positive_test * prior_disease) + (0.05 * (1 - prior_disease))  # Total probability of testing positive

posterior_disease_given_positive_test = bayesian_inference(prior_disease, likelihood_positive_test, evidence_positive_test)
print(f"Probability of having the disease given a positive test result: {posterior_disease_given_positive_test}")

#
```
